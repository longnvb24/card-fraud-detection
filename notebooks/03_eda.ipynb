## 1. Load the data and compute Fraud rate

```python
from pathlib import Path
import polars as pl
import plotly.express as px

SILVER_DIR = Path("..") / "data" / "silver"
df = pl.read_parquet(SILVER_DIR / "train.parquet")
```

```python
df
```

```python
fraud_rate = df.select(pl.col("isFraud").mean())
print(fraud_rate)
```

## 2. Transaction by Time

### 2.1 Transaction by Day

```python
# TransactionDT isn't a date, it's the number of seconds so this cell converts it into day number
df_day = df.with_columns(
    day = pl.col("TransactionDT") // 86400          # 86400 seconds in a day
)
```

```python
daily = df_day.group_by("day").agg(
    pl.len().alias("n"),
    pl.col("isFraud").mean().alias("fraud_rate"),
).sort("day")
daily
```

```python
px.line(daily, x="day", y="fraud_rate", title="Fraud rate per day")
```

```python
px.bar(daily, x="day", y="n", title="Number of transactions per day")
```

**Insight:** The training data covers 182 days. The fraud rate is not stable over time:
it is 2.5% in the first 30 days, rises to about 4.0% in days 31–120, then settles
at about 3.4% in days 121–180. Daily rates are noisy, ranging from 1.1% to 7.0%.
The first month also has more transactions (134K vs about 90K in later months).

**Implication:** Fraud patterns change over time, so a random train/test split would
mix past and future and give over-optimistic results. I will validate with a
time-based split (train on earlier days, test on the last period) and monitor
drift after deployment.

### 2.2. Transaction by Hour

```python
df_hour = df.with_columns(
    hour = (pl.col("TransactionDT") // 3600) % 24
)
```

```python
hourly = df_hour.group_by("hour").agg(
    pl.len().alias("n"),
    pl.col("isFraud").mean().alias("fraud_rate")
).sort("hour")
with pl.Config(tbl_rows=30):
    print(hourly)
```

```python
px.bar(hourly, x="hour", y="fraud_rate")
```

**Insight:** The fraud rate depends strongly on the hour of day. It rises from about
3% in the busy hours to a peak of 10.6% at hour 7, and stays above 5% from
hour 4 to hour 10. These are the quietest hours: hour 7 has only 3,704 transactions
vs about 42,000 at hour 19. Across the 24 hours, transaction volume and fraud rate
are strongly negatively correlated (r = −0.73). Note: TransactionDT has an unknown
start point, so "hour 7" may not be local time, but the pattern is still real.

**Implication:** Hour of day is a useful feature. Transactions in low-traffic hours
are about 2–3 times riskier, which makes "hour between 4 and 10" a candidate
baseline rule. These hours cover only about 7% of transactions, so a rule
based on them would not flood analysts with alerts.

## 3. Basic statistical description of Transaction Amount

```python
des_df = df.group_by("isFraud").agg(
    pl.len().alias("n"),
    pl.col("TransactionAmt").mean().alias("average"),
    pl.col("TransactionAmt").median().alias("median"),
    pl.col("TransactionAmt").quantile(0.9).alias("p90"),
    pl.col("TransactionAmt").quantile(0.99).alias("p99"),
)
print(des_df)
```

```python
df_amt = df.with_columns(log_amt = pl.col("TransactionAmt").log1p())
```

```python
px.histogram(df_amt, x="log_amt", color="isFraud", histnorm="probability density", barmode="overlay", nbins=100)
```

## 4. Transaction Amount by Band

```python
df_band = df.with_columns(amt_band = pl.col("TransactionAmt").bin_intervals(
    intervals=[25, 50, 100, 200, 500, 1000],
    labels=["<25", "25-50", "50-100", "100-200", "200-500", "500-1000", ">1000"]))
```

```python
band = df_band.group_by("amt_band").agg(
    pl.len().alias("n"),
    pl.col("isFraud").mean().alias("fraud_rate")
).sort("amt_band")
print(band)
```

**Insight:** Fraud and normal transactions have similar typical amounts
(median A$75 vs A$68.5). By amount band, the relationship is U-shaped rather than
increasing: the highest fraud rate is in the smallest band, below A$25 (7.0%, n = 43,329),
it drops to about 3% between A$25 and A$200, rises to 5.7% between A$500 and A$1,000,
then falls to 2.5% at A$1,000 and above.

**Implication:** "Flag large amounts" is a weak baseline rule on its own. The high
fraud rate in very small amounts is consistent with card testing (fraudsters making
small purchases to check that a stolen card works); I will check this in feature
engineering with card-level features such as the amount compared with the card's
usual amount. Because the relationship is non-linear, a tree-based model like
LightGBM should capture it better than a linear model.
